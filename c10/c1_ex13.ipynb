Use autograd to find the gradient of f(x,y)=sin(x^2y) at the point (x,y)=(1.2,3.4).

```python
import torch
```

```python
x = torch.tensor(1.2, requires_grad=True)
y = torch.tensor(3.4, requires_grad=True)
```

```python
f = ((x **2)*y).sin()
```

```python
f.backward()
```

```python
print(x.grad.item())
print(y.grad.item())
```
